```python
from math import sqrt
```

```python
class Rectangle:
    """Class that defines a rectangle shape"""
    def __init__(self, width: int, height: int) -> None:
        if not isinstance(width, int):
            raise TypeError("Width must be an integer value.")
        
        if width <= 0 :
            raise ValueError("Invalid value. Width should be a positive numerical value.")
       
        if not isinstance(height, int):
            raise TypeError("Height must be an integer value.")
        
        if height <= 0 :
            raise ValueError("Invalid value. Height should be a positive numerical value.")

        self.width  = width
        self.height = height


    def set_width(self, width: int) -> None:
        self.width = width

    def set_height(self, height: int) -> None:
        self.height = height
    
    def get_area(self) -> int:
        return self.width * self.height
    
    def get_perimeter(self) -> int:
        return 2 * (self.width + self.height)
    
    def get_diagonal(self) -> float:
        return sqrt(self.width ** 2 + self.height ** 2)
    
    def get_picture(self) -> str:
        picture = ""
        char    = "*"

        if self.width > 50 or self.height > 50:
            picture = "Too big for picture."
            return picture
        
        horizontal_line = char * self.width
            
        picture = (horizontal_line + "\n") * (self.height - 1) + horizontal_line

        return picture
    
    def get_amount_inside(self, rectangle) -> int:
        number_of_shapes = 0

        if rectangle.width > self.width or rectangle.height > self.height:
            return number_of_shapes
        
        num_width  = int(self.width / rectangle.width)
        num_height = int(self.height / rectangle.height)

        number_of_shapes = num_width * num_height
        
        return number_of_shapes


    # SETTERS AND GETTERS
    @property
    def width(self) :
        return self._width
    
    @width.setter
    def width(self, value) :
        if hasattr(self, '_width') and not isinstance(value, int):
            raise TypeError("Width must be an integer value.")
        
        if value <= 0 :
            raise ValueError("Invalid value. Width should be a positive numerical value.")
        
        self._width = value

    @property
    def height(self) :
        return self._height
    
    @height.setter
    def height(self, value) :
        if hasattr(self, '_height') and not isinstance(value, int):
            raise TypeError("Height must be an integer value.")
        
        if value <= 0 :
            raise ValueError("Invalid value. Height should be a positive numerical value.")
        
        self._height = value

    # INTERNAL / SPECIAL METHODS
    def __repr__(self):
        return f"Rectangle(width={self.width}, height={self.height})"
```

```python
class Square(Rectangle):
    """Class that defines a square"""
    def __init__(self, side: int) -> None:
        if not isinstance(side, int):
            raise TypeError("Side must be an integer value.")
        
        if side <= 0 :
            raise ValueError("Invalid value. Side should be a positive numerical value.")

        super().__init__(side, side)
        self.side = side


    def set_width(self, value: int) -> None:
        self.width  = value
        self.height = value
    
    def set_height(self, value: int) -> None:
        self.width  = value
        self.height = value

    def set_side(self, value: int) -> None:
        self.width  = value
        self.height = value

    # SETTERS AND GETTERS
    @property
    def side(self) :
        return self._width
    
    @side.setter
    def side(self, value) :
        if hasattr(self, '_side') and not isinstance(value, int):
            raise TypeError("Side must be an integer value.")
        
        if value <= 0 :
            raise ValueError("Invalid value. Side should be a positive numerical value.")
        
        self._side = value

    # INTERNAL / SPECIAL METHODS
    def __repr__(self):
        return f"Square(side={self.side})"
```

```python
# Rectangle tests
rect = Rectangle(4, 8)
print("Shape:", rect.get_picture(), sep='\n')
print(f"Perimeter: {rect.get_perimeter()}")
print(f"Area: {rect.get_area()}")
print(f"Amount of shapes that could fit inside: {rect.get_amount_inside(rect)}")
print(f"Instance as string: {repr(rect)}")

rect2 = Rectangle(width=10, height=5)
print(f"Instance as string: {repr(rect2)}")
rect2.set_height(3)
print("Another shape:", rect2.get_picture(), sep='\n')


print(f"Diagonal: {Rectangle(3, 6).get_diagonal()}")
```

```python
# Square tests
print(Square(side=4).get_picture())
Rectangle(15, 10).get_amount_inside(Square(5))
```
