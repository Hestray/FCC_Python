# Numerical operations
In Python, there is no increment (`++`) or decrement (`--`)

```python
print("Unary +")
num = 4
print("+num\t = ", +num)     # 4
print("-num\t =", -num)     # -4
print("--num\t = ", --num)    # 4

print("\nUnary -")
num = -4
print("num\t =", num)      # -4
print("+num\t =", +num)     # 4
print("++num\t =", ++num)    # 4
print("-num\t = ", -num)     # 4
```

```python
# variable definition
running_total   = 0

num_of_friends  = 4

appetizers      = 37.89
main_courses    = 57.34
desserts        = 39.39
drinks          = 64.21
```

```python
running_total += appetizers + main_courses + desserts + drinks
print(f"Total bill so far: {running_total}")
```

```python
tip = running_total * 0.25
print(f"Tip amount: {tip}")
```

```python
running_total += tip
print(f"Total with tip: {running_total}")
```

```python
final_bill  = running_total / num_of_friends
print(f"Bill per person: {final_bill}")

each_pays   = round(final_bill, 2)
print(f"Each person pays: {each_pays}")
```
