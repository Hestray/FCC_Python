# STRING MANIPULATION
Using:
- slicing
- `str.replace('')`
- `str.upper()`
- `str.lower()`
- `str.strip()` (~ trim)
- `str.split()`
- `''.join(iterable)`
- `str.startswith('')`
- `str.endswith('')`
- `str.count('')`
- `str.capitalize()`
- `str.isupper()`
- `str.islower()`
- `str.title()` (~ capitalize, but for every word)
where `str` refers to an user-defined string

Slicing has syntax `[start:stop:step]`.

```python
first_name = "John"
last_name  = "Doe"
print(first_name, last_name)
```

```python
full_name = first_name + last_name
print(full_name)

full_name = first_name + " " + last_name
print(full_name)
```

```python
address = "123 Main Street"
print(address)

address += ", Apartment 4B"
print(address)
```

```python
employee_age    = 28
employee_info   = full_name + ' is '
employee_info  += str(employee_age) + " years old"

print(employee_info)
```

```python
experience_years = 5
experience_info  = "Experience: " + str(experience_years) + " years"

print(experience_info)
```

```python
employee_card  = f"Employee: {full_name}"
employee_card += f" | Age: {employee_age}"

print(employee_card)
```

```python
position = "Data Analyst"
salary = 75000

employee_card += f" | Position: {position} | Salary: ${salary}"
print(employee_card)
```

```python
employee_code   = "DEV-2026-JD-001"

# slicing
department      = employee_code[0:3]
year_code       = employee_code[4:8]
initials        = employee_code[9:11]

last_three      = employee_code[-3:]

print(department)
print(year_code)
print(initials)

print(last_three)
```
