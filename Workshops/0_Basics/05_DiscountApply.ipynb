For the code below, we can see that if we don't use `nonlocal`, then that variable will not be modified by the `inner_func()`.

```python
def outer_func1():
    msg = 'Hello there!'
    res = "Initial res"  # Declare res in the enclosing scope

    def inner_func():
        res = 'How are you?' # essentially creates a temporary var
        print(res)
        print(msg)  # Accessing msg from outer_func()

    inner_func()
    print(res)  # Res is not modified

def outer_func2():
    msg = 'Hello there!'
    res = "Initial res"  # Declare res in the enclosing scope

    def inner_func():
        nonlocal res  # Allow modification of an enclosing variable
        res = 'How are you?'
        print(msg)  # Accessing msg from outer_func()

    inner_func()
    print(res)  # Now res is accessible and modified


print("outer_func1:")
outer_func1()

print("\n--------\nouter_func2:")
outer_func2()
```

Similarly, we could use `global` to a locally defined variable accessible globally.

# Discount App

```python
def apply_discount(price, discount) :
    if not (type(price) == int or type(price) == float) :
        return "The price should be a number"
    
    if not (type(discount) == int or type(discount) == float) :
        return "The discount should be a number"
    
    if price <= 0 :
        return "The price should be greater than 0"
    
    if discount < 0 or discount > 100 :
        return "The discount should be between 0 and 100"

    final_price = price * (1 - discount / 100)
    return final_price
```

```python
# test cases
print(f"{apply_discount("a", True)}")
print(f"{apply_discount(0, True)}")
print(f"{apply_discount(0, 0)}")
print(f"{apply_discount(10, 0)}")
print(f"{apply_discount(50, 20)}")

# checking results
print(f"{apply_discount(100, 20)}")
print(f"{apply_discount(200, 50)}")
print(f"{apply_discount(50, 0)}")
print(f"{apply_discount(500, 100)}")
print(f"{apply_discount(74.5, 20.0)}")
```
