# Note about iterables
Lists: `[]`

Sets: `{}`

Dictionaries: `{key: value}` (~ maps)

Tuple: `()`

# Lists methods
`A` refers to an user-defined list (`[el1, el2, el3]`):
- `A.append(el)`, where `el` can be another list or a value (eg. `[1, 2, 3, [4, 5]]`)
- `A.extend(el)` for `el` that is a another list (eg. `[1, 2, 3, 4, 5]`)
- `A.insert(index, el)`
- `A.remove(el)`
- `A.pop(index)`
- `A.clear()`
- `A.sort()`, which modifies `A`
- `sorted(A)`, which returns a new list instead of modifying `A`
- `A.reverse()`
- `A.index(el)`, which returns the first index where `el` is found, but `el` must exist in `A`

Assignment through the access operator is possible, unlike for tuples.

# Tuple methods
`T` refers to an user-defined tuple (`(el1, el2, el3)`).
- `T.count(el)`
- `T.index(el)`
- `sorted(T)` will create a new *list*;
    - `sorted(T, key = len)` sorts by the length
    - `sorted(T, reverse = True)` sorts in reverse

Destructuring:
```
developer = ('Alice', 34, 'Python')
name, *rest = developer
```

# Loop methods
- `range(start, stop, step)`, where `stop` is non-inclusive
- `enumerate(A)` returns both index and element
- `zip(A1, A2)`

````python
languages = ['Spanish', 'English', 'Russian', 'Chinese']
list_languages = list(enumerate(languages))
print(list_languages)   # [(0, 'Spanish'), (1, 'English'), (2, 'Russian'), (3, 'Chinese')]

print(f"enumerate(...) type: {type(enumerate(languages))}")         # enumerate
print(f"list_languages type: {type(list_languages)}")               # list
print(f"list_languages element type: {type(list_languages[0])}")    # tuple
````

````python
hobbies = ['Art', 'Programming', 'Reading', 'Playing the piano']
proficiency = ['Good', 'Good', 'Leisure', 'Good']

for hobby, pr in zip(hobbies, proficiency):
    print(f"Hobby: {hobby}")
    print(f"Proficiency: {pr}")
````

# List comprehension
The syntax is essentially `[expression condition for_loop_outer for_loop_inner condition == True]`. The way to think about it is from end to start.

If the condition is simple (eg. only one `if`, no `else`), then the condition is placed at the end: `[expression for_loop condition == True]`. In truth, this is the actual syntax, since `expression` includes the condition. Most of the time, it's simpler to just look at the errors and examples instead of knowing the exact syntax. Things can get confusing very quick

````python
numbers = [1, 2, 3, 4, 5]
result = [(num, 'Even') if num % 2 == 0 else (num, 'Odd') for num in numbers]
print(result)

# the following two are almost equal, except that by having the condition first, we need the if-else structure
print([num if num % 2 else "" for num in numbers])
print([num for num in numbers if num % 2])
````

````python
hobbies = [['Art', 'artistic'], 
           ['Programming', 'technical'], 
           ['Reading', 'none'],
           ['Playing the piano', 'artistic'],
           ['Playing the flute', 'artistic'],
           ['Walking', 'none'],
           ['Math problems', 'technical'],
           ['Taking care of plants', 'none']]

print([hobby[0] for hobby in hobbies])
# print([hobby[0] for hobby in hobbies if hobby[1] == 'artistic'] if hobby.count(' ') != 0) # THIS WILL RESULT IN AN ERROR: 'expected 'else' after 'if' expression`
print([hobby[0] for hobby in hobbies if hobby[1] == 'artistic' and hobby[0].count(' ') != 0])
print([hobby[0] if hobby[0].count(' ') != 0 else '' for hobby in hobbies if hobby[1] == 'artistic'])

print([1 for hobby in hobbies if hobby[1] == 'artistic' for hobby_name in hobby[0] if hobby_name.count(' ') != 0])
print("Number of empty spaces in artistic hobbies: ", sum([1 for hobby in hobbies if hobby[1] == 'artistic' for hobby_name in hobby[0] if hobby_name.count(' ') != 0]))
````

## `filter()` and `map()`
Takes a predicate (usually a function) and an iterable. Their function calls are identical, except that `filter` seems to return the element if the expression (function) is true, while `map` evaluates and returns the results as items. This is because `filter` creates a new list with all objects that return true when evaluates, and `map` simply applies a function across all elements. 

`lambda` functions shine when applied with those two functions.

````python
squared_even_numbers = list(filter(lambda x: x % 2 == 0, map(lambda x: x ** 2, range(10))))
print(*squared_even_numbers)
````
